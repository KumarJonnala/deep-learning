## Mean Squared Rrror

```python
def mean_squared_error(y, yhat):
    return np.mean((y - yhat)**2)
```

```python
mean_squared_error(1, 0.9997)
```

```python
mean_squared_error(1, 0.9)
```

```python
mean_squared_error(1,0.6)
```

```python
mean_squared_error(1,0.1192)
```

```python
mean_squared_error(np.array([1, 2, 3]), np.array([1, 2, 3]))
```

```python
mean_squared_error(np.array([1, 1, 1]), np.array([2, 2, 2]))
```

## Cross Entropy
Chief distintion is not the cost value but the rate of change of the cost values, especially if saturated neurons are involved\
Cross entropy cost is such that the dC/dw is proportional to the (y - yhat),\
i.e., larger the deviation, the greater is the rate of change of cost C.\
Hence larger deviation lead to faster the neuron learning.\
Therefore, it is commonly used cost function in NN

```python
def cross_entropy(y, a): # a is the activation function output of any neuron, a = yhat for output layer
    return - 1 * (y * np.log(a) + (1-y) * np.log(1-a))
```

```python
cross_entropy(1, 0.9997)
```

```python
cross_entropy(1, 0.9)
```

```python
cross_entropy(1, 0.6)
```

```python
cross_entropy(1, 0.1192)
```
