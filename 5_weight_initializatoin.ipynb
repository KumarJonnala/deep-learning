## Weight Initialization
Weight initialization approaches to examine their impact on layer outputs.

```python
import numpy as np
import matplotlib.pyplot as plt

import torch 
import torch.nn as nn
import torch.optim as optim
```

```python
# number of neurons
n_input = 784
n_dense = 256
```

```python
# weight and bias initialization
#w_init = nn.init.normal_ # Normal distribution initialization for weights
w_init = nn.init.xavier_normal_ # Xavier initialization for weights
b_init = nn.init.zeros_
```

```python
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(784, 64),
    nn.Sigmoid(),
    #nn.Tanh(),
    #nn.ReLU()
)

def init_weights(module):
    if isinstance(module, nn.Linear):
        w_init(module.weight) # apply to weights
        b_init(module.bias)   # apply to biases

model.apply(init_weights)     # walks every layer and calls init_weights on each
```

```python
# test with random input
x = np.random.rand(n_input)
```

```python
# forward pass
a = model(torch.from_numpy(x).float().unsqueeze(0))
```

```python
plt.hist(a.detach().numpy().flatten(), bins=30)
plt.title("Distribution of activations with Normal initialization")
plt.xlabel("Activation value")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(a.detach().numpy().flatten(), bins=30)
plt.title("Distribution of activations with Xavier initialization")
plt.xlabel("Activation value")
plt.ylabel("Frequency")
plt.show()
```

Neuron activations saturated when weights initialized with normal distribution\
Neuron activations close to normal when weights initialized with xavier distribution
