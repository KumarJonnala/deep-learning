# Activation Functions
The non-linear functions permit deep learning models to approximate any continious functions

# The Perceptron Activation Function
Output [0 or 1]

```python
def perceptron(z):
    return 1 if z >= 0 else 0
```

```python
z = [ -10, -1.2, 0, 2.45, 10]  # Result of (weights * inputs) + bias

for i in z:
    print(f"perceptron({i}) = {round(perceptron(i), 5)}")
```

# The Sigmoid Activation Function
Range [0,1]\
Sigmoid saturates at higher negative or positive values\
Negligible change in the output with subltle change in extreme z values

```python
from math import e
```

```python
def sigmoid(z):
    return 1/(1+e**-z)
```

```python
z = [-100, -10, -1, 0, 0.00001, 5, 10000]

for i in z:
    print(f"sigmoid({i}) = {round(sigmoid(i), 5)}")
```

# The Tanh Activation Function
Range [-1,1]\
Negative inputs -> negative activations and positive inputs -> positive activations\
Less likey to saturate as it is zero centered

```python
def tanh(z):
    return (e**z - e**-z) / (e**z + e**-z)
```

```python
z = [-100, -10, -1, 0, 0.00001, 5, 100]

for i in z:
    print(f"tanh({i}) = {round(tanh(i), 5)}")
```

# The ReLU Activation Function
Range [0, z]
Inputs z <=0 -> zero activation and z > 0, activation = z

```python
def relu(z):
    return max(0, z)
```

```python
z = [-10, -5, 0, 2, 10000]

for i in z:
    print(f"relu({i}) = {round(relu(i), 5)}")
```

## The Softmax Activation Function 
Range: [0,1]

```python
def softmax(z):
    exp_z = [e**i for i in z]
    sum_exp_z = sum(exp_z)
    return [round(i/sum_exp_z, 5) for i in exp_z]
```

```python
z = [-1, 1, 5] 

print(f"softmax({z}) = {softmax(z)}")
print(f"sum of softmax({z}) = {round(sum(softmax(z)), 5)}")
```
